What is DataWriter?

DataWriter is a class used to save data into different file formats.

Why do we use it?

After reading or processing data, we need to store the data in an output file.

```python
import csv
import json
import pickle
```

What is it?

  Saving a CSV file means writing data from Python into a .csv file.

Why do we use it?

  We use it to store tabular data in rows and columns so it can be used later.

```python
class DataWriter:

    # Save CSV file
    def write_csv(self, data, file_path):
        with open(file_path, "w", newline="") as file:
            writer = csv.writer(file)
            writer.writerows(data)
```

What is it?

   Saving a JSON file means writing data from Python into a .json file.

Why do we use it?

   We use it to store structured data in JSON format. JSON is commonly used for data exchange between applications.

```python
    # Save JSON file
    def write_json(self, data, file_path):
        with open(file_path, "w") as file:
            json.dump(data, file, indent=4)
```

What is it?

  Saving a Pickle file means saving a Python object into a .pkl file.

Why do we use it?

   We use Pickle when we want to store Python data/objects and load them back later without converting

```python
    # Save Pickle file
    def write_pickle(self, data, file_path):
        with open(file_path, "wb") as file:
            pickle.dump(data, file)
```

What is it?

  Saving a TXT file means writing data from Python into a .txt file.

Why do we use it?

  We use it to store data as simple plain text, which is easy to read

```python
    # Save TXT file
    def write_txt(self, data, file_path):
        with open(file_path, "w") as file:
            file.write(str(data))
```
