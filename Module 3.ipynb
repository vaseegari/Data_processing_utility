```python
import logging
from pathlib import Path
```

1. class LoggerManager:
 
    Definition:
   
        LoggerManager is a class used to configure and manage application logging.

          

```python
class LoggerManager:

    def __init__(self):

        # Create logs folder
        log_directory = Path("logs")
        log_directory.mkdir(parents=True, exist_ok=True)

        # Create logger
        self.logger = logging.getLogger("application")
        self.logger.setLevel(logging.INFO)

        # Create log file
        file_handler = logging.FileHandler(
            "logs/application.log",
            encoding="utf-8"
        )

        # Set log format
        formatter = logging.Formatter(
            "%(asctime)s - %(levelname)s - %(message)s"
        )

        file_handler.setFormatter(formatter)

        # Connect handler to logger
        self.logger.addHandler(file_handler)

    def get_logger(self):
        return self.logger
```

1. logger_manager = LoggerManager()

What: 

  Creates an object of the LoggerManager class.

Why: 

  When we create this object, the __init__() method runs automatically and sets up:
  
2. logger = logger_manager.get_logger()

What:

  Calls the get_logger() method.

Why: 

  It gets the configured logger so we can actually write messages.

```python
logger_manager = LoggerManager()
logger = logger_manager.get_logger()
```



logger.info() 

   This line records an INFO-level log message.
   
   This is used to record normal application activities. Here, it records that the application has started.


```python
logger.info("Application started")
```

 This line records a WARNING-level log message.
