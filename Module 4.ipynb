1.Try → Used to run code that may cause an error.

2.Except → Used to handle the error if an error occurs in the try block.

3.Raise → Used to manually trigger an exception, especially a custom exception.

4.Finally → Used for code that must run whether an error occurs or not.

```python
class InvalidFileException(Exception):
    pass


def read_file(file_path):

    try:
        file = open(file_path, "r")
        data = file.read()
        print("File read successfully")
        return data

    except FileNotFoundError:
        raise InvalidFileException("File not found")

    finally:
        print("File operation completed")
```
